# CyberShield

## Database Connectivity

```python
import sqlite3
import pandas as pd
```

## Load Engineered Dataset

```python
data = pd.read_csv("../output/engineered_dataset.csv")

print(data.shape)

data.head()
```

## Create SQLite Database

```python
connection = sqlite3.connect("../database/cybershield.db")

cursor = connection.cursor()

print("Database Connected Successfully")
```

## Store Dataset in Database

```python
data.to_sql(
    "network_data",
    connection,
    if_exists="replace",
    index=False
)

print("Dataset Stored Successfully")
```

## Display Available Tables

```python
query = """
SELECT name
FROM sqlite_master
WHERE type='table';
"""

pd.read_sql(query, connection)
```

## Retrieve Dataset

```python
query = """
SELECT *
FROM network_data
LIMIT 10;
"""

pd.read_sql(query, connection)
```

## Retrieve Attack Records

```python
query = """
SELECT *
FROM network_data
WHERE label = 1
LIMIT 10;
"""

pd.read_sql(query, connection)
```

## Retrieve Normal Traffic

```python
query = """
SELECT *
FROM network_data
WHERE label = 0
LIMIT 10;
"""

pd.read_sql(query, connection)
```

## Count Normal and Attack Records

```python
query = """
SELECT
label,
COUNT(*) AS Total
FROM network_data
GROUP BY label;
"""

pd.read_sql(query, connection)
```

## Count Attack Categories

```python
query = """
SELECT
attack_cat,
COUNT(*) AS Total
FROM network_data
GROUP BY attack_cat
ORDER BY Total DESC;
"""

pd.read_sql(query, connection)
```

## Create Incident Log Table

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS incident_log (

incident_id INTEGER PRIMARY KEY AUTOINCREMENT,

attack_type TEXT,

threat_level TEXT,

recommended_action TEXT,

status TEXT,

timestamp DATETIME DEFAULT CURRENT_TIMESTAMP

)
""")

connection.commit()

print("Incident Log Table Created")
```

## View Incident Log Table

```python
query = """
SELECT *
FROM incident_log;
"""

pd.read_sql(query, connection)
```

## Close Database Connection

```python
connection.close()

print("Database Closed Successfully")
```
