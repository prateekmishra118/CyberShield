## Rank Models by F1 Score

```python
final_results = final_results.sort_values(
    by="F1 Score",
    ascending=False
).reset_index(drop=True)

final_results
```

## Statistical Summary of Model Performance

```python
metric_columns = [
    "Accuracy",
    "Precision",
    "Recall",
    "F1 Score"
]
statistical_summary = final_results[metric_columns].agg(
    ["mean", "std", "min", "max"]
)
statistical_summary
```

## Best Model by Evaluation Metric

```python
for metric in metric_columns:
    best_index = final_results[metric].idxmax()
    best_model = final_results.loc[best_index, "Model"]
    best_score = final_results.loc[best_index, metric]
    print(
        f"{metric}: {best_model} ({best_score:.4f})"
    )
```

## Performance Improvement After Hyperparameter Tuning

```python
tuning_comparison = pd.DataFrame({
    "Version": [
        "Before Tuning",
        "After Tuning"
    ],
    "Accuracy": [
        0.897091,
        0.898915
    ],
    "Precision": [
        0.876863,
        0.881563
    ],
    "Recall": [
        0.896951,
        0.895342
    ],
    "F1 Score": [
        0.886793,
        0.888399
    ]
})
tuning_comparison
```

## Calculate Tuning Improvements

```python
before = tuning_comparison.iloc[0]
after = tuning_comparison.iloc[1]
print("Hyperparameter Tuning Improvement")
print("----------------------------------")
for metric in metric_columns:
    improvement = after[metric] - before[metric]
    percentage = (improvement / before[metric]) * 100
    print(
        f"{metric}: "
        f"{improvement:+.6f} "
        f"({percentage:+.3f}%)"
    )
```

## Result Interpretation and Discussion

The final evaluation shows that HistGradientBoosting achieved the best overall balance of performance among the evaluated models, with an initial F1 score of 0.886793. After hyperparameter tuning, the F1 score improved to 0.888399, indicating a measurable improvement in the balance between precision and recall.

The tuned model achieved an accuracy of approximately 0.8989 and precision of approximately 0.8816. Although recall decreased slightly after tuning, the overall F1 score improved, indicating that the tuned model provides a better overall classification balance.

For cybersecurity detection, reducing missed attacks is particularly important. Therefore, the protection threshold was reduced from 0.50 to 0.30. This reduced false negatives from 1,366 to 385 and increased recall from approximately 89.53% to 97.05%. The trade-off was an increase in false positives from 1,570 to 3,098.

The attack-category analysis identified Exploits and Fuzzers as the two largest categories in the evaluated traffic. The protection engine assigned actions according to threat severity, with ISOLATE assigned to Critical threats, RESTRICT assigned to Medium threats, and BLOCK assigned to High threats.

Overall, the results demonstrate that the CyberShield pipeline combines machine-learning classification with a threshold-based protection mechanism to prioritize the detection of potentially harmful traffic.

## Discussion of Protection Threshold Selection

The protection threshold of 0.30 was selected because the primary objective of the CyberShield detection stage is to minimize missed attacks. At the default threshold of 0.50, the model produced 1,366 false negatives. Reducing the threshold to 0.30 decreased false negatives to 385 and increased recall from 89.53% to 97.05%.

However, this improvement resulted in an increase in false positives from 1,570 to 3,098. Therefore, the threshold represents a security-oriented trade-off: the system accepts more false alarms in exchange for detecting a larger proportion of potentially malicious traffic.

For a cybersecurity protection system, this trade-off can be appropriate because missed attacks may have a greater operational impact than additional alerts requiring investigation.

## Conclusion

The CyberShield machine-learning pipeline successfully integrates data preprocessing, feature engineering, model training, model comparison, hyperparameter tuning, threat detection, attack classification, and protection-action assignment.

Among the evaluated models, HistGradientBoosting provided the best overall F1 performance. Hyperparameter tuning produced a small improvement in F1 score from 0.886793 to 0.888399.

The protection threshold analysis demonstrated that lowering the detection threshold from 0.50 to 0.30 substantially reduced false negatives and increased recall to approximately 97.05%. The protection engine then mapped detected threats to appropriate actions based on their severity.

The final evaluation therefore demonstrates the feasibility of combining machine-learning-based threat detection with an automated protection policy in the CyberShield framework.

## Limitations

1. The evaluation is based on the available dataset and may not represent all types of real-world network traffic.

2. The protection effectiveness results are based on the implemented CyberShield protection logic and should not be interpreted as proof of real-world attack prevention.

3. The threshold selection involves a trade-off between false negatives and false positives. A lower threshold can improve attack detection while generating more alerts.

4. The current protection engine uses predefined severity-to-action rules rather than a dynamic response policy.

5. Model performance may change when the system encounters previously unseen attack patterns or significantly different network traffic.
