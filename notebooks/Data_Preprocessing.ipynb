# Data Preprocessing

## Step 11: Check Missing Values

```python
# Count missing values in each column
data.isnull().sum()
```

## Step 12: Duplicate Record Analysis

```python
# Check duplicate records

duplicate_records = data.duplicated().sum()

print("Total Duplicate Records:", duplicate_records)
```

## Step 13: Investigate Duplicate Records

```python
# Display the first few duplicate rows
duplicates = data[data.duplicated()]
duplicates.head()
```

```python
# Display the shape of duplicate records
duplicates.shape
```

```python
# Check duplicate percentage
duplicate_percentage = (duplicate_records / len(data)) * 100

print(f"Duplicate Percentage: {duplicate_percentage:.2f}%")
```

## Step 14: Remove Duplicate Records

```python
# Remove duplicate records

data = data.drop_duplicates()

print("Dataset Shape After Removing Duplicates:", data.shape)
```

```python
# Verify duplicate removal

print("Remaining Duplicate Records:", data.duplicated().sum())
```

## Step 15: Verify Dataset Shape After Cleaning

```python
print("Final Dataset Shape:", data.shape)
```

## Step 16: Data Type Verification

```python
# Check data types of all columns

data.dtypes
```

```python
# Count different data types

data.dtypes.value_counts()
```

```python
# Display categorical columns

categorical_columns = data.select_dtypes(include='object').columns

print("Categorical Columns:")

categorical_columns
```

```python
# Display numerical columns

numerical_columns = data.select_dtypes(include=['int64','float64']).columns

print("Number of Numerical Columns:", len(numerical_columns))

numerical_columns
```

## Step 17: Label Encoding of Categorical Features

```python
from sklearn.preprocessing import LabelEncoder
```

```python
label_encoder = LabelEncoder()
```

```python
categorical_columns
```

```python
for column in categorical_columns:
    data[column] = label_encoder.fit_transform(data[column])
```

```python
data.dtypes.value_counts()
```

```python
data.head()
```

## Step 18: Feature Scaling (Standardization)

```python
from sklearn.preprocessing import StandardScaler
```

```python
# Create StandardScaler object

scaler = StandardScaler()
```

```python
# Create feature dataset by excluding target columns

features = data.drop(columns=['label', 'attack_cat'])

target = data[['label', 'attack_cat']]
```

```python
# Scale only feature columns

scaled_features = pd.DataFrame(
    scaler.fit_transform(features),
    columns=features.columns
)
```

```python
# Combine scaled features with target columns

processed_data = pd.concat(
    [scaled_features, target.reset_index(drop=True)],
    axis=1
)
```

```python
processed_data.head()
```

```python
processed_data.describe()
```

## Step 19: Save Processed Dataset

```python
import os

# Create output directory if it doesn't exist
os.makedirs("../output", exist_ok=True)

# Save processed dataset
processed_data.to_csv("../output/processed_dataset.csv", index=False)

print("Processed dataset saved successfully.")
```
