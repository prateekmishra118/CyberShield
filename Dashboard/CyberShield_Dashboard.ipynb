```python
tuning_comparison = pd.DataFrame({
    "Version": [
        "Before Tuning",
        "After Tuning"
    ],
    "Accuracy": [
        0.897091,
        0.898915
    ],
    "Precision": [
        0.876863,
        0.881563
    ],
    "Recall": [
        0.896951,
        0.895342
    ],
    "F1 Score": [
        0.886793,
        0.888399
    ]
})

tuning_comparison
```

```python
before = tuning_comparison.iloc[0]
after = tuning_comparison.iloc[1]

print("Hyperparameter Tuning Improvement")
print("----------------------------------")

for metric in [
    "Accuracy",
    "Precision",
    "Recall",
    "F1 Score"
]:

    change = after[metric] - before[metric]

    percentage = (
        change / before[metric]
    ) * 100

    print(
        f"{metric}: "
        f"{change:+.6f} "
        f"({percentage:+.3f}%)"
    )
```

```python
threshold_analysis = pd.DataFrame({
    "Configuration": [
        "Default Threshold (0.50)",
        "Protection Threshold (0.30)"
    ],
    "False Negatives": [
        1366,
        385
    ],
    "False Positives": [
        1570,
        3098
    ],
    "Recall": [
        0.895341710082746,
        0.9705026049647564
    ]
})

threshold_analysis["Recall (%)"] = (
    threshold_analysis["Recall"] * 100
)

threshold_analysis
```

```python
default_fn = 1366
protection_fn = 385

fn_reduction = default_fn - protection_fn

fn_reduction_percentage = (
    fn_reduction / default_fn
) * 100

print("False-Negative Analysis")
print("-----------------------")
print("Default Threshold FN :", default_fn)
print("Protection Threshold FN :", protection_fn)
print("False Negatives Reduced :", fn_reduction)
print(
    f"Reduction Percentage : "
    f"{fn_reduction_percentage:.2f}%"
)
```
